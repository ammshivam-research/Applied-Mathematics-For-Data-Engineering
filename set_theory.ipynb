# **SET THEORY**

**Question 1** : How do elements inside a set stay unique while sharing a connection, and what are the different ways they can be similar to one another?

A set manages the uniqueness of its present elements.

But how?

By removing the duplicates from the set i.e. any element with more than 1 multiplicity must be removed except the source element.
```
E.g. {1,2,3,1,2,3,1,1,4}  # Not a true mathematical set

{1,2,3,4} # Convert it into a true mathematical set
```

Unique elements within a set shows its belongingness either through container-based or rule-based commonality or both.

Container-based commonality happens simply because the items live together inside the same set, like putting a number and a letter in the same box.

Rule-based commonality happens when the items share a specific property or has extracted after applying a condition. E.g : 1, 2, 5, and 10 all being factors of 10. so every element in the {1,2,3,4,5}, are considered to have the a rule based commonality.

**Question 4** : Define Cardinality within sets.

The **cardinality** of a set is the number of **distinct elements** it contains.  
It is commonly denoted by \( |A| \) or \( n(A) \).

**Example 1: Simple Finite Set**

Let  
$$ A = \{2, 5, 7, 11, 13\} $$

**Steps:**
1. List the elements: \( 2, 5, 7, 11, 13 \)
2. Count the distinct elements → there are **5** elements.

Result:
$$ |A| = 5 $$

**Example 2: Set with Duplicate Elements**

Let  
$$ B = \{a, b, a, c, b, d\} $$

**Important Note:**
A set only keeps **unique** elements. After removing duplicates:

$$ B = \{a, b, c, d\} $$

**Result:**
$$ |B| = 4 $$

# **Russel Paradox**
### What is the problem?

In old set theory, people believed:

> “You can make a set of **anything** you want.”

Russell showed this idea is broken.

---

### The Strange Set

Imagine we make a set called **R**.

**R** contains **only** those sets that do **not** contain themselves.

Now ask a simple question:

**Does R contain itself?**

---

### Two possibilities – both fail

**Possibility 1:**  
R contains itself.  
But the rule says R only contains sets that do **not** contain themselves.  
So R should **not** contain itself.  
→ Contradiction.

**Possibility 2:**  
R does **not** contain itself.  
But the rule says that any set that does not contain itself **must** be in R.  
So R **should** contain itself.  
→ Contradiction.

---

### What does this mean?

We cannot make such a set.  
The idea “you can make a set of anything” is wrong.

This is called **Russell’s Paradox**.

---

### Simple Real-Life Analogy

Think of a barber who says:

> “I shave **only** those men who do not shave themselves.”

Question: Does the barber shave himself?

- If yes → he is shaving someone who shaves himself → against his rule.
- If no → he is a man who does not shave himself → so he should shave himself.

Same problem.

---

### Final Point

This paradox forced mathematicians to create stricter rules about how sets can be formed (modern set theory).

# **Possible Relationship Among Sets**


```
Parent Set = {1,2,3,4,5}
```
**Scenario 1** :              
```
A = {1,2,3} ; B = {1,2,3}

Result : Equal Sets (A = B) , Improper Set (A ⊆ B)

(A = B) meaning : (A ⊆ B) AND ((B ⊆ A)
```
**Scenario 2** :
```
A = {1,2,3} ; B = {4,5,6}

Result : Disjoint Set (A ≠ B) , (A ⊄ B), (A ∩ B = ϕ)
```
**Scenario 3** :             

```
A = {1,2,3} ; B = {1,2,3,4}

Result : Proper Set (A ⊂ B), (A ⊊ B), (B ⊃ A)

(B ⊃ A) means B is a proper subset of A.
```
**Scenario 4** :       

```
A = {1,4} ; B = {1,2,3,5}

Result: Intersecting or Overlapping Set (A ∩ B ≠ ϕ), (A ⊄ B)
```
